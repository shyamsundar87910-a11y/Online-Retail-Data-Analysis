##  Extracting Year and Month from Invoice Date
---

The `.dt` accessor is used to extract date-related information from the InvoiceDate column.

- **Year:** Extracts the year from each transaction date.
- **Month:** Extracts the numerical month (1–12).
- **MonthName:** Extracts the full month name.

These columns help perform monthly and yearly sales trend analysis.

```python
df_clean['Year'] = df_clean['InvoiceDate'].dt.year

df_clean['Month'] = df_clean['InvoiceDate'].dt.month

df_clean['MonthName'] = df_clean['InvoiceDate'].dt.month_name()
```

##  Monthly Sales Analysis
---

The `groupby()` function groups the dataset by Year, Month, and MonthName.

The `sum()` function calculates the total revenue for each month using the TotalSales column.

The `reset_index()` method converts the grouped result into a DataFrame.

This analysis helps identify monthly revenue patterns and understand sales performance over time.

```python
monthly_sales = df_clean.groupby(
    ['Year', 'Month', 'MonthName']
)['TotalSales'].sum().reset_index()

monthly_sales
```

## Visualizing Monthly Revenue
---

The `plt.figure()` function sets the size of the chart.

The `sns.barplot()` function creates a bar chart to visualize monthly revenue.

- **X-axis:** Month names.
- **Y-axis:** Total revenue.
- **Title:** Monthly Revenue Analysis.

The `plt.xticks(rotation=45)` function rotates the month labels for better readability.

This visualization helps compare revenue across different months.

```python
plt.figure(figsize=(14, 6))

sns.barplot(
    data=monthly_sales,
    x='MonthName',
    y='TotalSales'
)

plt.title('Monthly Revenue Analysis')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

## Identifying the Highest Revenue Month
---

The `idxmax()` function returns the index of the row containing the maximum TotalSales value.

The `loc[]` accessor retrieves the complete row using that index.

This analysis helps identify the month with the highest total revenue.

```python
monthly_sales.loc[monthly_sales['TotalSales'].idxmax()]
```

## Country-wise Sales Analysis
---

The `groupby('Country')` function groups transactions based on country.

The `sum()` function calculates the total revenue generated by each country.

The `sort_values()` method sorts countries in descending order of TotalSales.

The `head(10)` function displays the top 10 countries by revenue.

This analysis helps identify the countries contributing the highest revenue to the online retail business.

```python
country_sales = df_clean.groupby('Country')['TotalSales'].sum().reset_index()

country_sales = country_sales.sort_values(
    by='TotalSales',
    ascending=False
)

country_sales.head(10)
```

##  Visualizing Top 10 Countries by Revenue
---

The `country_sales.head(10)` function selects the top 10 countries based on total revenue.

The `sns.barplot()` function creates a horizontal bar chart to compare revenue across these countries.

- **X-axis:** Total Revenue.
- **Y-axis:** Country.
- **Title:** Top 10 Countries by Revenue.

This visualization helps identify and compare the highest revenue-generating countries.

```python
top_countries = country_sales.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_countries,
    x='TotalSales',
    y='Country'
)

plt.title('Top 10 Countries by Revenue')
plt.xlabel('Total Revenue')
plt.ylabel('Country')
plt.show()
```

### Daily Sales Analysis
---

The `.dt.date` accessor extracts the date from the InvoiceDate column.

The `groupby('Date')` function groups transactions by date.

The `sum()` function calculates the total revenue generated on each day.

The `reset_index()` method converts the grouped result into a DataFrame.

## Visualizing Daily Sales Trend
---

The `plt.plot()` function creates a line chart to visualize daily revenue over time.

- **X-axis:** Date.
- **Y-axis:** Total Revenue.
- **Title:** Daily Sales Trend.

Additional functions:
- `plt.grid(True)` adds grid lines.
- `plt.xticks(rotation=45)` rotates date labels.
- `plt.tight_layout()` adjusts the chart layout.

This visualization helps identify daily sales fluctuations and understand revenue trends over time.

```python
plt.figure(figsize=(14, 6))

plt.plot(
    daily_sales['Date'],
    daily_sales['TotalSales']
)

plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Weekday-wise Sales Analysis
